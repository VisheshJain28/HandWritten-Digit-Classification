```python
import tensorflow
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense,Flatten
```

```python
(x_train,y_train),(x_test,y_test) = keras.datasets.mnist.load_data()
```

```python
x_train.shape
```

```python
x_train[0]
```

```python
y_train
```

```python
import matplotlib.pyplot as plt
plt.imshow(x_train[0])
```

```python
#to make all value between 0 and 1

x_train = x_train/255
x_test = x_test/255
```

```python
model = Sequential()

#Flatten the data to make it one dimensional i.e (1,784)
model.add(Flatten(input_shape=(28,28)))

model.add(Dense(128,activation='relu'))
model.add(Dense(10,activation='softmax'))
```

```python
model.summary()
```

```python
model.compile(loss='sparse_categorical_crossentropy',optimizer='Adam')
```

```python
hist = model.fit(x_train,y_train,epochs=25,validation_split=0.2)
```

```python
#Predict the probability of every number i.e. 0,1,2,3...9 for each test data

y_prob = model.predict(x_test)
```

```python
#The maximum prob of that is the output

y_pred = y_prob.argmax(axis=1)
y_pred
```

```python
from sklearn.metrics import accuracy_score

accuracy_score(y_pred,y_test)
```

```python
plt.plot(hist.history['loss'])
plt.plot(hist.history['val_loss'])
```
